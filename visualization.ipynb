```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
employee_name = ["Bob", "Rajat", "Gaurishankar", "Dutch"]
salary = [200, 2003, 2050, 35555]
bonus = [5000, 10000, 2555, 3666]
```

```python
# Line Plot
plt.figure(figsize=(6, 4))
plt.plot(
    employee_name, salary, marker="o", color="Black", linestyle="solid", label="salary"
)
plt.plot(employee_name, bonus, color="green", marker="o", label="bonus", alpha=0.6)
plt.xlabel("Employee_name", color="red")
plt.ylabel("Salary / Bonus", color="Red")
plt.title("Total Salary Distribution by Employees", color="Red")
plt.grid(alpha=1, linestyle="dotted")
plt.legend()
plt.savefig("salary_bonus.png")
# alpha means opacity
plt.show()
```

```python
# Line Plot
plt.figure(figsize=(6, 4))
plt.plot(employee_name, bonus, marker="o", color="Black", linestyle="solid")
plt.xlabel("Employee_name", color="red")
plt.ylabel("Salary", color="Red")
plt.title("Total Salary and bonus Distribution by Employees", color="Red")
plt.grid(alpha=1, linestyle="dotted")
# alpha means opacity
plt.show()
```

```python
##BAR CHART
```

```python
plt.figure(figsize=(6, 4))
plt.bar(
    employee_name,
    salary,
    edgecolor="black",
    color=["red", "black", "orange", "green"],
    width=0.52,
)
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.title("Total Salary Distribution of Employees")
plt.xticks(rotation=15)
# horizontal allignment =left,right,center, vertical allignment=top/bottom/center
for idx, value in enumerate(salary):
    plt.text(idx, value, f"Rs{value}", ha="center", va="bottom")
plt.show()
```

```python
df = pd.read_csv("hr_cleaned_data.csv")
df.head()
```

```python
perf_emp = df.groupby("PerformanceScore")["EmpID"].count().reset_index()
perf_emp
```

```python
# pie chart
plt.pie(perf_emp["EmpID"], labels=perf_emp["PerformanceScore"], autopct="%1.1f%%",explode=(0.3,0,0,0))
plt.title("% of Employees by Performance")

plt.show()
```

```python
# Donut chart
plt.pie(
    perf_emp["EmpID"],
    labels=perf_emp["PerformanceScore"],
    autopct="%1.1f%%",
    # width edge properties
    wedgeprops={"width": 0.4,'edgecolor':'black'},
    pctdistance=0.8,
    explode=(0,00.2,0,0)
)
plt.title("% of Employees by Performance")

plt.show()
```
