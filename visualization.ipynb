```python
import pandas as pd
import matplotlib.pyplot  as plt
```

```python
employee_name=['Bob','Rajat','Gaurishankar','Dutch']
salary=[200,2003,2050,35555]
```

```python
#Line Plot
plt.figure(figsize=(6,4))
plt.plot(employee_name,salary,marker='o',color='Black',linestyle='solid')
plt.xlabel('Employee_name',color='red')
plt.ylabel('Salary',color='Red')
plt.title('Total Salary Distribution by Employees',color='Red')
plt.grid(alpha=1,linestyle='dotted')
#alpha means opacity
plt.show()
```
