```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
employee_name = ["Bob", "Rajat", "Gaurishankar", "Dutch"]
salary = [200, 2003, 2050, 35555]
bonus = [5000, 10000, 2555, 3666]
```

```python
# Line Plot
plt.figure(figsize=(6, 4))
plt.plot(
    employee_name, salary, marker="o", color="Black", linestyle="solid", label="salary"
)
plt.plot(employee_name, bonus, color="green", marker="o", label="bonus", alpha=0.6)
plt.xlabel("Employee_name", color="red")
plt.ylabel("Salary / Bonus", color="Red")
plt.title("Total Salary Distribution by Employees", color="Red")
plt.grid(alpha=1, linestyle="dotted")
plt.legend()
plt.savefig("salary_bonus.png")
# alpha means opacity
plt.show()
```

```python
# Line Plot
plt.figure(figsize=(6, 4))
plt.plot(employee_name, bonus, marker="o", color="Black", linestyle="solid")
plt.xlabel("Employee_name", color="red")
plt.ylabel("Salary", color="Red")
plt.title("Total Salary and bonus Distribution by Employees", color="Red")
plt.grid(alpha=1, linestyle="dotted")
# alpha means opacity
plt.show()
```

```python
##BAR CHART
```

```python
plt.figure(figsize=(6, 4))
plt.bar(
    employee_name,
    salary,
    edgecolor="black",
    color=["red", "black", "orange", "green"],
    width=0.52,
)
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.title("Total Salary Distribution of Employees")
plt.xticks(rotation=15)
# horizontal allignment =left,right,center, vertical allignment=top/bottom/center
for idx, value in enumerate(salary):
    plt.text(idx, value, f"Rs{value}", ha="center", va="bottom")
plt.show()
```

```python
df = pd.read_csv("hr_cleaned_data.csv")
df.head()
```

```python
perf_emp = df.groupby("PerformanceScore")["EmpID"].count().reset_index()
perf_emp
```

```python
# pie chart
plt.pie(perf_emp["EmpID"], labels=perf_emp["PerformanceScore"], autopct="%1.1f%%",explode=(0.3,0,0,0))
plt.title("% of Employees by Performance")

plt.show()
```

```python
# Donut chart
plt.pie(
    perf_emp["EmpID"],
    labels=perf_emp["PerformanceScore"],
    autopct="%1.1f%%",
    # width edge properties
    wedgeprops={"width": 0.4,'edgecolor':'black'},
    pctdistance=0.8,
    explode=(0,00.2,0,0)
)
plt.title("% of Employees by Performance")

plt.show()
```

```python
plt.hist(df['Salary'],edgecolor='black')
plt.xlabel('Salary')
plt.ylabel('Frequency / Number of Employees')
plt.show()
```

```python
mean_salary=df['Salary'].mean()
median_salary=df['Salary'].median()
```

```python
plt.figure(figsize=(6,4))
# kde -> Kernal density Estimator
sns.histplot(df['Salary'],bins=10,kde=True,color='blue')
# axvline -> axis vertical line
plt.axvline(mean_salary,color='red',label='mean_salary')
plt.axvline(median_salary,color='green',label='median_salary')
plt.legend()
plt.show()
```

Column Chart

```python
# df.columns
#Find total employees by each department and gender

dep_gen=df.groupby(['Department','Gender'])['EmpID'].count().reset_index().rename(
   columns= {'EmpID':'Total Employees'}
)
dep_gen
```

```python
'''
-hue ->used to create column chart
'''

plt.figure(figsize=(10,4))
ax=sns.barplot(data=dep_gen,x='Department',y='Total Employees',hue='Gender',errorbar=None)
for container in ax.containers:
    plt.bar_label(container)
plt.show()
```

Sub-Plots

```python
dep_details=df.groupby('Department').agg(
    Total_Employees=('EmpID','count'),
    Total_Salary=('Salary','sum'),
    Total_Absences=('Absences','sum'),
    Total_Projects=('SpecialProjectsCount','sum')
).reset_index()
dep_details.columns=dep_details.columns.str.replace('_',' ',regex=False)
dep_details
```

```python
fig,axes=plt.subplots(2,2,figsize=(14,10))
plt.subplots_adjust(hspace=0.5)

ax1 = axes[0,0]
ax1.bar(dep_details['Department'],dep_details['Total Salary'])
ax1.set_xlabel('Department')
ax1.set_ylabel('Total Salary')
ax1.set_title('Total Salary distribution on Each Department')
ax1.tick_params(axis='x',rotation=15)

ax2 = axes[0,1]
ax2.bar(dep_details['Department'],dep_details['Total Absences'],edgecolor='Black',color='red')
ax2.set_xlabel('Department')
ax2.set_ylabel('Total Absences')
ax2.set_title('Total Absences of Each Department')
ax2.tick_params(axis='x',rotation=15)
```
