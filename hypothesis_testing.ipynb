Hypothesis Testing

```python
''' 1 Year business
average profit = 50000 -> Sample

Total Sales = 1300000 -> Population

Types of Hypothesis Testing
1. T Test
        a.One Sample T Test
        b.Two Sample T Test

2.ANOVA Test 
3.Chi Square Test


--Statistics 
--Probability

stats->(Actual Mean - Sample Mean)/(sd/root(total_number))

-If Stats value increases P-Value decrease
-If Stats value decreases P-Value increases

'''
```

```python
#Statistical calculation -> scipy -> scientific python
from scipy import stats
import pandas as pd
```

```python
df=pd.read_csv('Dataset/pizza_sales.csv')
df.head()
```

One Sample T test

is the average total price of all pizza sales different from $15?

```python
'''
import numpy as np
actual_mean = population.mean()
sd=population.std()
total_data = np.sqrt(48619)

top_val = actual_mean - sample
bot_val = sd / total_data

top_val / bot_val

'''
```

```python
alpha=0.05
sample = 15
population = df['total_price']
```

```python
# stats.ttest_1samp(population,popmean=sample)
statistics,p_value=stats.ttest_1samp(population,popmean=sample)
```

```python
ho = f'The average total price of all pizza sales equals to ${sample}'
ha = f'The average total price of all pizza sales is not equal to ${sample}'
```

```python
def decision_function(ha,ho):
    if p_value < alpha :
        print(f'Decision: Failed to accept null hypothesis.\nConclusion: Needs to accept alternate hypothesis.\n Ha: {ha}')
    else:
       print(f'Decision: Failed to reject null hypothesis.\nConclusion: Needs to accept Null hypothesis.\n Ho: {ho}')
```

```python
decision_function(ha,ho)
```

Two Sample T Test

The average unit price of Classic Pizzas and Chicken Pizzas Same?

```python
alpha = 0.5
classic_price = df[df['pizza_category'] == 'Classic']['unit_price']
chicken_price = df[df['pizza_category']=='Chicken']['unit_price']
```

```python
#stats.ttest_ind(category1, category2 , equal_var=True/False)
t_statistics,p_value =stats.ttest_ind(classic_price,chicken_price,equal_var=False)
```

```python
ho = 'The Average unit price of Classic Pizzas and Chicken Pizzas are equal.'
ha= 'The average unit price of Classic and Chicken Pizzas are not Equal'
```

```python
decision_function(ho=ho,ha=ha)
```

ANOVA Test (Analysis Of Variance)

Do the average total_price generated by all four pizza categories
(Classic, Chicken, Supreme, and Veggie) differ significantly from one another?​

```python
alpha = 0.5
categories = ['Classic', 'Chicken', 'Supreme', 'Veggie' ]
grouped_prices = []

for category in categories:
   price=df[df['pizza_category']==category]['total_price']
   grouped_prices.append(price)
```

```python
t_statistics, p_value = stats.f_oneway(*grouped_prices)
```

```python
#pizza_price=df.groupby(['pizza_category'])['total_price'].mean().rename_axis('Pizza Category').rename('Total Price').map("${:,.2f}".format).to_markdown(numalign='center',stralign='center')
```

```python
ho='The average price generated by all four pizza categories are equal to one another.'
ha='The average  total price generated by all four pizza categories are not equal to one another.'

print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
decision_function(ho=ho,ha=ha)
```

### Chi-Squared Test

IS pizza Category and Pizza Size are dependent to one another

```python
alpha=0.5
# contingency analysis -> crosstab(Cross Tabulation)
contingency =  pd.crosstab(df['pizza_category'],df['pizza_size'])
contingency
```

```python
#dof = (numbers of rows - 1 ) * (number of columns - 1)
```

```python
statistics,p_value,dof,ef=stats.chi2_contingency(contingency)
```

```python
ho = 'Pizza Category and pizza size are dependent to one other.'
ha = 'Pizza Category and pizza size are independent to one other.'

decision_function(ho=ho,ha=ha)
```
